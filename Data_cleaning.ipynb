# 1. Sales.csv

```python
# Detecting nulls 
Sales.isnull().sum()
```

# 4. Products

```python
Products = pd.read_csv("Raw_Dataset/Products.csv")
```

```python
Products.head(5)
```

```python
Products.sample(5)
```

```python
Products.info()
```

```python
# Renaming columns
Products = Products.rename(columns={
    'ProductKey' : 'product_key',
    'Product Name':'product_name',
    'Brand': 'brand',
    'Color':'color',
    'Unit Cost USD':'unit_cost_usd',
    'Unit Price USD':'unit_price_usd',
    'SubcategoryKey':'subcategory_key',
    'Subcategory':'subcategory',
    'CategoryKey':'category_key',
    'Category':'category'
})
```

```python
Products.columns
```

```python
Products.dtypes
```

```python
# Removing Special characters from column
Products['unit_cost_usd'] = Products['unit_cost_usd'].str.replace(',','')
Products['unit_cost_usd'] = Products['unit_cost_usd'].str.replace('$','',regex=False)

Products['unit_price_usd'] = Products['unit_price_usd'].str.replace(',','')
Products['unit_price_usd'] = Products['unit_price_usd'].str.replace('$','',regex=False)
```

```python
Products.head()
```

```python
# Fixing data types
Products['unit_cost_usd']  =  Products['unit_cost_usd'].astype('float')
Products['unit_price_usd']  =  Products['unit_price_usd'].astype('float')
```

```python
Products.head(5)
```

```python
Products.dtypes
```

```python
Products.describe().round(1)
```

```python
Products.sample(10)
```
